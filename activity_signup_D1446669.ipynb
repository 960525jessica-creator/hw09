```python
# === 測試一：原始資料 ===
activity_names = ("桌遊體驗", "電影欣賞")
board_activity, movie_activity = activity_names

board_raw = " Amy, Ben, amy, Cora "
movie_raw = " ben, Dora, ERIC, ben "

unit_fees = {"桌遊體驗": 120, "電影欣賞": 80}

# 移除半形空白、轉小寫、依逗號拆分，保留原始順序與重複項目
board_list = board_raw.replace(" ", "").lower().split(",")
movie_list = movie_raw.replace(" ", "").lower().split(",")

# 轉成 set 去重並做集合運算
board_set = set(board_list)
movie_set = set(movie_list)

both_set = board_set & movie_set          # 共同參加者（交集）
all_set = board_set | movie_set           # 全部參加者（聯集）
only_board_set = board_set - movie_set    # 只參加桌遊者（差集：桌遊 - 電影）
only_movie_set = movie_set - board_set    # 只參加電影者（差集：電影 - 桌遊）

# 排序成 list 以便顯示，並用切片取得前兩位通知預覽
board_sorted = sorted(board_set)
movie_sorted = sorted(movie_set)
both_sorted = sorted(both_set)
only_board_sorted = sorted(only_board_set)
only_movie_sorted = sorted(only_movie_set)
all_sorted = sorted(all_set)
notify_preview = all_sorted[:2]

# 計算人數、人次與應收費用（金額維持數值）
board_count = len(board_set)
movie_count = len(movie_set)
both_count = len(both_set)
unique_total = len(all_set)
total_signups = board_count + movie_count

board_fee = board_count * unit_fees[board_activity]
movie_fee = movie_count * unit_fees[movie_activity]
total_fee = board_fee + movie_fee

summary = {
    "桌遊人數": board_count,
    "電影人數": movie_count,
    "共同人數": both_count,
    "總不重複人數": unique_total,
    "報名人次": total_signups,
    "桌遊應收": board_fee,
    "電影應收": movie_fee,
    "總應收": total_fee,
}

# 輸出完整摘要
print("HW09 活動報名摘要（測試一：原始資料）")
print(f"活動：{board_activity}，每人 {unit_fees[board_activity]:.2f} 元")
print(f"活動：{movie_activity}，每人 {unit_fees[movie_activity]:.2f} 元")
print(f"桌遊整理串列：{board_list}")
print(f"電影整理串列：{movie_list}")
print(f"桌遊去重名單：{board_sorted}")
print(f"電影去重名單：{movie_sorted}")
print(f"桌遊人數：{summary['桌遊人數']} 人")
print(f"電影人數：{summary['電影人數']} 人")
print(f"共同參加者：{both_sorted}")
print(f"共同人數：{summary['共同人數']} 人")
print(f"只參加桌遊者：{only_board_sorted}")
print(f"只參加電影者：{only_movie_sorted}")
print(f"全部參加者：{all_sorted}")
print(f"總不重複人數：{summary['總不重複人數']} 人")
print(f"報名人次：{summary['報名人次']} 人次")
print(f"通知名單前兩位：{notify_preview}")
print(f"桌遊應收：{summary['桌遊應收']:.2f} 元")
print(f"電影應收：{summary['電影應收']:.2f} 元")
print(f"總應收：{summary['總應收']:.2f} 元")
```
